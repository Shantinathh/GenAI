```python
from langchain_core.documents import Document
from langchain_community.vectorstores import Chroma
from dotenv import load_dotenv
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
load_dotenv()

docs = [
    Document(page_content="Python is widely used in Artificial Intelligence.", metadata={"source": "AI_book"}),
    Document(page_content="Pandas is used for data analysis in Python.", metadata={"source": "DataScience_book"}),
    Document(page_content="Neural networks are used in deep learning.", metadata={"source": "DL_book"}),
]
```

```python
embedding_model = GoogleGenerativeAIEmbeddings(model='gemini-embedding-2-preview')

embeddings = Chroma.from_documents(
    embedding=embedding_model,
    documents=docs,
    persist_directory="Chroma-db"
)
```

```python
result = embeddings.similarity_search("What is used for data analysis?",k=2)

for i in result:
    print(i)
```

```python
result = embeddings.similarity_search("Where python is used widely?",k=1)
for i in result:
    print(i.page_content)
    print(i.metadata)
```
